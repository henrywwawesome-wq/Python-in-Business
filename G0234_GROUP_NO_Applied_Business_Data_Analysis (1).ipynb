````python
# ============================================================
# STEP 3: READ DATASET
# Purpose: Download the dataset from Kaggle and load it into
#          a pandas DataFrame for inspection and cleaning.
# ============================================================

# --- 3.1 Download the dataset from Kaggle ---
# kagglehub.dataset_download() connects to Kaggle, downloads all files
# for the given dataset slug, and saves them to a local cache folder.
# It returns the path string of the folder where files are stored.
# Syntax: path = kagglehub.dataset_download('owner/dataset-name')
path = kagglehub.dataset_download("prince7489/e-commerce-sales")
print(f"Dataset downloaded to: {path}")

# --- 3.2 See what files were downloaded ---
# os.listdir() returns a list of all filenames in the given folder.
# Syntax: os.listdir(folder_path)
files = os.listdir(path)
print(f"Files found: {files}")
````

````python
# --- 3.3 Load the CSV file into a pandas DataFrame ---

# Use a list comprehension to find the first CSV file automatically.
# This avoids hardcoding the filename and works even if the name changes.
# List comprehension syntax: [item for item in list if condition]
csv_file = [f for f in files if f.endswith('.csv')][0]

# os.path.join() safely combines folder path + filename into one full path.
# Syntax: os.path.join(folder, filename)
filepath = os.path.join(path, csv_file)
print(f"Loading file: {filepath}")

# pd.read_csv() reads the CSV and returns it as a pandas DataFrame.
# encoding='utf-8'  → handles standard characters correctly
# low_memory=False  → avoids mixed-type warnings on larger files
df = pd.read_csv(filepath, encoding='utf-8', low_memory=False)

print('=' * 50)
print(' DATASET LOADED SUCCESSFULLY')
print('=' * 50)
# .shape returns a tuple: (number_of_rows, number_of_columns)
print(f"Shape: {df.shape[0]} rows x {df.shape[1]} columns")
````

````python
# --- 3.4 Preview the first 5 rows ---
# .head(n) shows the first n rows of the DataFrame.
# Use this to confirm the data looks correct after loading.
# Syntax: df.head(5)
print("First 5 rows:")
df.head(5)
````

````python
# --- 3.5 Inspect column names and data types ---
# .dtypes shows each column and its data type.
# Common types: int64 (whole numbers), float64 (decimals), object (text/mixed)
# Syntax: df.dtypes
print("Column Data Types:")
print('-' * 40)
print(df.dtypes)
````

````python
# --- 3.6 Check for missing values ---
# .isnull() returns a True/False DataFrame (True where value is NaN).
# .sum() counts the True values per column = number of missing entries.
# Syntax: df.isnull().sum()
print("Missing Values Per Column:")
print('-' * 40)
missing     = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

# pd.DataFrame() builds a new table from a dictionary of columns
missing_table = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})

# Only display rows that actually have missing values
has_missing = missing_table[missing_table['Missing Count'] > 0]
if len(has_missing) > 0:
    print(has_missing)
else:
    print("No missing values found in the dataset.")
````

````python
# ============================================================
# --- 3.7 Basic Preprocessing ---
# Purpose: Clean the dataset so it is ready for EDA and ML.
# Each step below includes its reason and syntax.
# ============================================================

# Step A: Strip whitespace from column names
# Reason: Accidental spaces in column names (e.g. ' Sales' vs 'Sales')
#         cause KeyError when you try to access df['Sales'].
# Syntax: df.columns = df.columns.str.strip()
df.columns = df.columns.str.strip()
print("Column names after stripping whitespace:")
print(list(df.columns))

# Step B: Remove duplicate rows
# Reason: Duplicate rows inflate counts and skew averages, which
#         produces misleading summaries and biased ML models.
# .duplicated().sum() counts rows that are exact copies of another row.
# .drop_duplicates() removes them, keeping only the first occurrence.
# Syntax: df.drop_duplicates(inplace=True)
print(f"\nDuplicate rows found   : {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)
print(f"Duplicates after removal: {df.duplicated().sum()}")

# Step C: Drop rows where essential columns are missing
# Reason: Rows without Sales, Quantity, Category, or Segment
#         cannot contribute to EDA or ML — they must be removed.
# Syntax: df.dropna(subset=['col1', 'col2'], inplace=True)
key_cols = ['Sales', 'Quantity', 'Category', 'Segment']
# List comprehension: only include columns that actually exist in df
key_cols_exist = [c for c in key_cols if c in df.columns]
df.dropna(subset=key_cols_exist, inplace=True)
print(f"\nRows after dropping NaN in key columns: {df.shape[0]}")

# Step D: Convert numerical columns to correct numeric dtype
# Reason: CSV files often read numbers as text (object dtype),
#         which prevents mathematical operations like mean() or sum().
# errors='coerce' → turns unconvertible values into NaN instead of crashing.
# Syntax: pd.to_numeric(df['col'], errors='coerce')
for col in ['Sales', 'Quantity', 'Discount', 'Profit']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
print("\nNumerical columns converted: Sales, Quantity, Discount, Profit")

# Step E: Convert date columns to datetime format
# Reason: Dates loaded as plain text cannot be sorted or used
#         for time-based grouping (e.g. month, year).
# Syntax: pd.to_datetime(series, errors='coerce')
for date_col in ['Order Date', 'Ship Date']:
    if date_col in df.columns:
        df[date_col] = pd.to_datetime(df[date_col], errors='coerce')
        print(f"Converted '{date_col}' to datetime format.")

# Step F: Reset the row index after dropping rows
# Reason: After removing rows, the index has gaps (e.g. 0, 2, 5...).
#         reset_index() renumbers rows cleanly from 0, 1, 2, 3...
#         drop=True discards the old index rather than adding it as a column.
# Syntax: df.reset_index(drop=True, inplace=True)
df.reset_index(drop=True, inplace=True)

print(f"\nPreprocessing complete.")
print(f"Final dataset: {df.shape[0]} rows x {df.shape[1]} columns")
````

````python
# ============================================================
# STEP 4.1 – SUMMARY STATISTICS
# Purpose: Get a numerical overview of all columns to understand
#          the range, average, and spread of the data.
# ============================================================

# .describe() automatically computes: count, mean, std, min, 25%, 50%, 75%, max
# for every numerical column in the DataFrame.
# Syntax: df.describe().round(decimal_places)
print("Summary Statistics – All Numerical Columns:")
print('-' * 55)
print(df.describe().round(2))
print()


# .value_counts() counts how many times each unique value appears.
# Syntax: df['column'].value_counts()
print("Segment Distribution:")
print(df['Payment Mode'].value_counts())

print()
print("Category Distribution:")
print(df['Category'].value_counts())

print()
print("Region Distribution:")
print(df['Region'].value_counts())
````

````python
# ============================================================
# STEP 4.2 – FUNCTIONS (Modular Code Design)
# Purpose: Define reusable functions that keep code organised
#          and avoid repeating the same logic multiple times.
# ============================================================

# --- FUNCTION 1: Classify sales into business performance tiers ---
# Uses: if-elif-else (three-way conditional)
# Syntax to define: def function_name(parameter): ... return value
def classify_sales(amount):
    """
    Classifies a sales value into Low, Medium, or High tier.
    Parameter: amount (float) - the sales dollar value
    Returns  : str - 'Low', 'Medium', or 'High'
    """
    # if-elif-else: checks each condition top to bottom;
    # only the first True block executes.
    if amount < 100:
        return 'Low'
    elif amount < 500:
        return 'Medium'
    else:
        return 'High'

# .apply() calls the function on every single row in the column.
# Syntax: df['new_column'] = df['existing_column'].apply(function_name)
df['Sales_Tier'] = df['Sales'].apply(classify_sales)
print("Function 1 applied — 'Sales_Tier' column created:")
print(df['Sales_Tier'].value_counts())
print()

# --- FUNCTION 2: Flag whether an order is profitable ---
# Uses: if-else (two-way conditional)
def is_profitable(profit):
    """
    Returns 1 if the order made a profit, 0 if it made a loss.
    Parameter: profit (float) - the profit value
    Returns  : int - 1 (profit) or 0 (loss)
    """
    # if-else: exactly one of two outcomes
    if profit > 0:
        return 1
    else:
        return 0

df['Is_Profitable'] = df['Profit'].apply(is_profitable)
print("Function 2 applied — 'Is_Profitable' column created:")
print(df['Is_Profitable'].value_counts())
print()

# --- FUNCTION 3: Print statistics for any numeric column using numpy ---
# Uses: if statement as a safety guard
def column_summary(dataframe, col_name):
    """
    Prints mean, median, std, min, and max for a numeric column.
    Parameters: dataframe (DataFrame), col_name (str)
    Returns   : None (prints results directly)
    """
    # if: prevent crash if column does not exist
    if col_name not in dataframe.columns:
        print(f"Column '{col_name}' not found.")
        return

    # .dropna() removes NaN values before computation
    # .to_numpy() converts pandas Series to a numpy array for fast math
    # Syntax: series.dropna().to_numpy()
    data = dataframe[col_name].dropna().to_numpy()

    print(f"  Column  : {col_name}")
    print(f"  Mean    : {np.mean(data):.2f}")
    print(f"  Median  : {np.median(data):.2f}")
    print(f"  Std Dev : {np.std(data):.2f}")
    print(f"  Min     : {np.min(data):.2f}")
    print(f"  Max     : {np.max(data):.2f}")
    print()

print("Function 3 — column_summary results:")
print('-' * 40)
column_summary(df, 'Sales')
column_summary(df, 'Profit')
column_summary(df, 'Quantity')
column_summary(df, 'Discount')
````

````python
# ============================================================
# STEP 4.3 – LOOPS (Automating Repetitive Tasks)
# Purpose: Use loops to process multiple groups or columns
#          without duplicating code manually.
# ============================================================

# --- LOOP 1: for loop — sales summary per Category ---
# .unique() returns all distinct values in the column as an array.
# Syntax: df['col'].unique()
categories = df['Category'].unique()

print("Sales Summary Per Category (for loop):")
print('-' * 55)
# for loop: runs the block once for each item in the iterable.
# Syntax: for variable in iterable:
for cat in categories:
    # Boolean filter: keep only rows where Category matches current cat
    # Syntax: df[df['col'] == value]
    subset = df[df['Category'] == cat]
    count  = len(subset)
    avg    = subset['Sales'].mean()
    total  = subset['Sales'].sum()
    print(f"  {cat}: {count} orders | Avg: ${avg:.2f} | Total: ${total:.2f}")

print()

# --- LOOP 2: for loop — IQR outlier detection on numeric columns ---
print("Outlier Detection Using IQR (for loop):")
print('-' * 55)
numeric_cols = ['Sales', 'Profit', 'Quantity', 'Discount']

for col in numeric_cols:
    # if: skip column if it does not exist in this dataset
    if col not in df.columns:
        continue
    # .quantile(q) returns the value at percentile q
    # Syntax: df['col'].quantile(0.25) → Q1 (25th percentile)
    Q1  = df[col].quantile(0.25)
    Q3  = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR   # lower fence: below this = outlier
    upper = Q3 + 1.5 * IQR   # upper fence: above this = outlier
    # | is the OR operator for combining pandas boolean conditions
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"  '{col}': {len(outliers)} outliers (range: {lower:.2f} to {upper:.2f})")

print()

# --- LOOP 3: for loop — average profit per region ---
print("Average Profit Per Region (for loop):")
print('-' * 55)
for region in df['Region'].unique():
    # .loc[condition, col] filters rows AND selects a specific column
    # Syntax: df.loc[df['col'] == val, 'target_col'].mean()
    avg_profit = df.loc[df['Region'] == region, 'Profit'].mean()
    # if-else inside loop: handle potential NaN gracefully
    if not np.isnan(avg_profit):   # np.isnan() → True if value is NaN
        print(f"  {region}: ${avg_profit:.2f}")
    else:
        print(f"  {region}: No data")

print()

# --- LOOP 4: while loop — count orders with profit above $100 ---
print("Orders With Profit > $100 (while loop):")
print('-' * 55)
profits      = df['Profit'].dropna().to_numpy()
i            = 0    # index — tracks current position
high_profit  = 0    # accumulator — counts qualifying orders

# while loop: keeps running as long as condition remains True.
# Syntax: while condition:
while i < len(profits):
    if profits[i] > 100:   # if: check each value
        high_profit += 1   # increment when condition met
    i += 1                 # always advance the index to avoid infinite loop

print(f"  Orders with profit > $100: {high_profit} ({high_profit/len(profits)*100:.1f}% of all orders)")
````

````python
# ============================================================
# STEP 4.4 – CONDITIONAL STATEMENTS
# Purpose: Use if / if-else / if-elif to translate data findings
#          into actionable business decisions.
# ============================================================

# --- Example 1: if (single condition) ---
# Alert if any orders resulted in a loss (Profit < 0)
loss_orders = len(df[df['Profit'] < 0])

# if: the block only runs when the condition is True
if loss_orders > 0:
    print(f"ALERT: {loss_orders} loss-making orders detected.")
    print("  Action: Review discount policy — high discounts may be causing losses.\n")

# --- Example 2: if-else (two-way decision) ---
# Evaluate whether average discount is too high
avg_discount = df['Discount'].mean()

# if-else: exactly one of the two blocks will execute
if avg_discount > 0.2:
    print(f"Average discount: {avg_discount:.1%} — Discounts are HIGH.")
    print("  Recommendation: Reduce discount rates to protect profit margins.\n")
else:
    print(f"Average discount: {avg_discount:.1%} — Discounts are within acceptable range.")
    print("  Recommendation: Maintain current pricing strategy.\n")

# --- Example 3: if-elif-else (multi-way decision) ---
# Assess overall business health using profit margin percentage
total_sales  = df['Sales'].sum()
total_profit = df['Profit'].sum()
margin       = (total_profit / total_sales) * 100

# if-elif-else: checks conditions in order; only the first True block runs
if margin >= 15:
    health = 'Excellent'
    action = 'Scale top-performing categories and regions.'
elif margin >= 5:
    health = 'Healthy'
    action = 'Look for cost reduction opportunities.'
elif margin >= 0:
    health = 'Marginal'
    action = 'Urgently review discount and shipping policies.'
else:
    health = 'Loss-Making'
    action = 'Immediate operational review required.'

print(f"Total Sales    : ${total_sales:,.2f}")
print(f"Total Profit   : ${total_profit:,.2f}")
print(f"Profit Margin  : {margin:.2f}%")
print(f"Business Health: {health}")
print(f"Recommendation : {action}")
````

````python
# ============================================================
# STEP 4.5 – COLUMN REMOVAL AND DATA TYPE CHANGES
# Purpose: Remove columns that are not useful for analysis and
#          convert data types to ensure correct results.
# ============================================================

# Dictionary mapping column name → reason for removal.
# Using a dictionary makes the justification easy to document and review.
drop_map = {
    'Row ID'     : 'Auto-generated row number — no business meaning',
    'Order ID'   : 'Unique order identifier — not a useful ML feature',
    'Customer ID': 'Unique customer identifier — not a useful ML feature',
    'Postal Code': 'Too granular; Region already captures location information',
    'Customer Name': 'Personal name — not relevant to sales prediction'
}

# for loop + if: drop each column only if it actually exists
for col, reason in drop_map.items():
    if col in df.columns:        # if: guard before dropping to prevent KeyError
        # Syntax: df.drop(columns=['col'], inplace=True)
        df.drop(columns=[col], inplace=True)
        print(f"Dropped '{col}': {reason}")
    else:
        print(f"Skipped '{col}': column not present in this dataset")

print()

# Convert text columns to pandas 'category' dtype
# Reason: 'category' dtype uses significantly less memory than 'object' (text),
#         and speeds up groupby, value_counts, and filtering operations.
# Syntax: df['col'] = df['col'].astype('category')
cat_cols = ['Category', 'Sub-Category', 'Region', 'Segment', 'Ship Mode']
for col in cat_cols:
    if col in df.columns:
        df[col] = df[col].astype('category')
        print(f"Converted '{col}' to category dtype")

print(f"\nFinal dataset: {df.shape[0]} rows x {df.shape[1]} columns")
print("\nUpdated column data types:")
print(df.dtypes)
````

## STEP 5 – Hypothesis Testing and Machine Learning Analysis

All calculations are built manually using **only numpy and pandas** — no sklearn or scipy needed.

| Task | Method | Tool |
|------|--------|------|
| Hypothesis Testing | Manual Welch's t-test | numpy |
| Classification | Logistic Regression via gradient descent | numpy |
| Regression | Linear Regression via Normal Equation | numpy |

### Key numpy Syntax for Step 5
| Function | Syntax | Purpose |
|----------|--------|---------|
| Mean | `np.mean(arr)` | Average |
| Sample variance | `np.var(arr, ddof=1)` | Variance for t-test |
| Square root | `np.sqrt(val)` | Used in t-test formula |
| Sigmoid | `1 / (1 + np.exp(-z))` | Logistic function |
| Dot product | `np.dot(A, B)` | Matrix multiplication |
| Column stack | `np.column_stack([a, b, c])` | Build feature matrix |
| Least squares | `np.linalg.lstsq(X, y, rcond=None)` | Linear regression weights |
| Conditional | `np.where(cond, val_T, val_F)` | Vectorised if-else |

````python
# ============================================================
# STEP 5A: HYPOTHESIS TESTING (Manual Welch's T-Test)
# Purpose: Test whether Consumer and Corporate segments have
#          significantly different average Sales values.
#
# Welch's t-test formula (does not assume equal variance):
#   t  = (mean1 - mean2) / sqrt(var1/n1 + var2/n2)
#   df = (var1/n1 + var2/n2)^2 / ((var1/n1)^2/(n1-1) + (var2/n2)^2/(n2-1))
# ============================================================

print('=' * 55)
print('HYPOTHESIS TESTING: Consumer vs Corporate Sales')
print('=' * 55)

print("""
H0 (Null Hypothesis)       : No significant difference in average
                             Sales between Consumer and Corporate segments.
H1 (Alternative Hypothesis): A significant difference EXISTS between
                             Consumer and Corporate average Sales.
Significance level (alpha) : 0.05
""")

# Separate into two groups using pandas boolean indexing
# Syntax: df.loc[df['col'] == value, 'target'].dropna().to_numpy()
consumer  = df.loc[df['Segment'] == 'Consumer',  'Sales'].dropna().to_numpy()
corporate = df.loc[df['Segment'] == 'Corporate', 'Sales'].dropna().to_numpy()

print(f"Consumer  — n={len(consumer)},  mean=${np.mean(consumer):.2f},  std=${np.std(consumer):.2f}")
print(f"Corporate — n={len(corporate)}, mean=${np.mean(corporate):.2f}, std=${np.std(corporate):.2f}")

# --- Compute t-statistic manually using numpy ---
n1, n2 = len(consumer), len(corporate)
mean1  = np.mean(consumer)
mean2  = np.mean(corporate)
# ddof=1 → sample variance (divides by n-1 instead of n)
var1   = np.var(consumer,  ddof=1)
var2   = np.var(corporate, ddof=1)

# t-statistic: difference in means divided by the standard error
t_stat = (mean1 - mean2) / np.sqrt(var1/n1 + var2/n2)

# Welch-Satterthwaite degrees of freedom
num   = (var1/n1 + var2/n2) ** 2
denom = ((var1/n1)**2 / (n1-1)) + ((var2/n2)**2 / (n2-1))
df_w  = num / denom

# Approximate p-value using the complementary error function
# Valid for large samples (n > 30); avoids needing scipy
from math import erfc
p_value = erfc(abs(t_stat) / np.sqrt(2))

print(f"\nT-Statistic          : {t_stat:.4f}")
print(f"Degrees of freedom   : {df_w:.1f}")
print(f"P-Value              : {p_value:.6f}")

# --- Decision using if-else ---
alpha = 0.05
print("\nDecision:")
if p_value < alpha:
    print(f"  p ({p_value:.6f}) < alpha ({alpha})  REJECT H0")
    print("  Conclusion: Significant difference in Sales between Consumer and Corporate.")
    print("  Business Insight: The two segments buy differently in terms of order size.")
    print("  Recommendation: Use separate pricing and promotion strategies per segment.")
else:
    print(f"  p ({p_value:.6f}) >= alpha ({alpha})  FAIL TO REJECT H0")
    print("  Conclusion: No significant difference found.")
    print("  Recommendation: A single unified pricing strategy can be applied to both segments.")
````

````python
# ============================================================
# STEP 5B: MACHINE LEARNING — LOGISTIC REGRESSION (numpy only)
# Purpose: Predict whether an order is from a Consumer (1)
#          or Corporate (0) customer.
#
# Method: Gradient Descent minimises Binary Cross-Entropy Loss
# Key formulas:
#   Sigmoid  : sigma(z) = 1 / (1 + e^(-z))   → maps score to probability
#   Gradient : (1/n) * X.T dot (predictions - true_labels)
#   Update   : weights = weights - lr * gradient
#
# Why Logistic Regression?
#   The target (Segment) is a binary category — classification is
#   the correct ML approach. Logistic Regression is simple, fast,
#   interpretable, and works well for linearly separable classes.
# ============================================================

print('=' * 55)
print('ML MODEL 1: LOGISTIC REGRESSION (numpy from scratch)')
print('=' * 55)

# --- 5B.1 Prepare binary classification data ---
ml = df[df['Segment'].isin(['Consumer', 'Corporate'])].copy()

# Encode target label: Consumer → 1,  Corporate → 0
# np.where(condition, value_if_true, value_if_false)
ml['label'] = np.where(ml['Segment'] == 'Consumer', 1, 0)

# Encode categorical features as integers using pd.factorize()
# pd.factorize(series)[0] → returns an array of integer codes
ml['cat_code'] = pd.factorize(ml['Category'])[0]
ml['reg_code'] = pd.factorize(ml['Region'])[0]
ml = ml.dropna(subset=['Sales', 'Quantity', 'Discount'])

# --- 5B.2 Build feature matrix X and target vector y ---
# np.column_stack() joins multiple 1D arrays into one 2D matrix
# Each array becomes one column (feature) in the matrix
# Syntax: np.column_stack([arr1, arr2, arr3])
X_raw = np.column_stack([
    ml['Sales'].to_numpy(),
    ml['Quantity'].to_numpy(),
    ml['Discount'].to_numpy(),
    ml['cat_code'].to_numpy(),
    ml['reg_code'].to_numpy()
])
y = ml['label'].to_numpy().astype(float)

# --- 5B.3 Normalise features to mean=0, std=1 ---
# Reason: Without normalisation, Sales (hundreds) dominates
#         Discount (0–1) and makes gradient descent unstable.
# Syntax: (array - np.mean(array, axis=0)) / np.std(array, axis=0)
X_mean = np.mean(X_raw, axis=0)        # column-wise mean
X_std  = np.std(X_raw,  axis=0)
X_norm = (X_raw - X_mean) / (X_std + 1e-8)  # +1e-8 prevents division by zero

# Add a bias column (column of 1s) so the model learns an intercept term
# np.hstack() joins arrays horizontally (side by side)
X = np.hstack([np.ones((X_norm.shape[0], 1)), X_norm])

# --- 5B.4 Manual 80/20 train/test split ---
# np.random.seed(42) → fixes random state so results are reproducible
# np.random.permutation() → randomly shuffles indices
np.random.seed(42)
idx   = np.random.permutation(len(X))
split = int(0.8 * len(X))              # 80% cutoff point
X_train, X_test = X[idx[:split]], X[idx[split:]]
y_train, y_test = y[idx[:split]], y[idx[split:]]
print(f"Training samples: {len(X_train)} | Testing samples: {len(X_test)}")

# --- 5B.5 Sigmoid function ---
# Converts any real number to a probability between 0 and 1.
# np.clip() prevents overflow when z is very large or very small.
# Syntax: 1 / (1 + np.exp(-z))
def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -500, 500)))

# --- 5B.6 Train via Gradient Descent ---
# np.zeros(n) → initialise all weights to 0
weights = np.zeros(X_train.shape[1])
lr      = 0.1     # learning rate — controls the step size per update
epochs  = 1000    # number of full passes through the training data

for epoch in range(epochs):
    # Forward pass: compute predicted probabilities
    # np.dot(A, b) → matrix-vector multiplication
    z     = np.dot(X_train, weights)
    y_hat = sigmoid(z)
    # Gradient: direction to move weights to reduce the loss
    grad  = np.dot(X_train.T, (y_hat - y_train)) / len(y_train)
    # Gradient descent update: move weights opposite to gradient
    weights = weights - lr * grad

print("Logistic Regression training complete.")

# --- 5B.7 Predict and evaluate ---
probs  = sigmoid(np.dot(X_test, weights))
# np.where: if probability >= 0.5 → predict 1 (Consumer), else 0 (Corporate)
y_pred = np.where(probs >= 0.5, 1, 0)

# Accuracy: proportion of correct predictions out of all predictions
accuracy = np.mean(y_pred == y_test)
print(f"\nModel Accuracy: {accuracy * 100:.2f}%")

# Manual Precision, Recall, F1 using numpy boolean comparisons
TP = np.sum((y_pred == 1) & (y_test == 1))  # correctly predicted Consumer
FP = np.sum((y_pred == 1) & (y_test == 0))  # predicted Consumer but actually Corporate
FN = np.sum((y_pred == 0) & (y_test == 1))  # predicted Corporate but actually Consumer
precision = TP / (TP + FP) if (TP + FP) > 0 else 0
recall    = TP / (TP + FN) if (TP + FN) > 0 else 0
f1        = (2 * precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print(f"Precision : {precision:.4f}   (of all predicted Consumer, how many were correct)")
print(f"Recall    : {recall:.4f}   (of all actual Consumer, how many did we catch)")
print(f"F1 Score  : {f1:.4f}   (balance of Precision and Recall)")

print("""
Business Interpretation:
  This model predicts whether an incoming order is from a Consumer or Corporate customer.
  Integrating this into the order system allows the business to:
  - Automatically route orders to the correct sales team
  - Apply segment-specific pricing and communication strategies
  - Prioritise high-value Corporate orders for faster fulfilment
""")
````

````python
# ============================================================
# STEP 5C: MACHINE LEARNING — LINEAR REGRESSION (numpy only)
# Purpose: Predict the Sales amount from Quantity, Discount, Category.
#
# Method: Normal Equation — exact closed-form solution
# Formula: weights = (X^T X)^-1 X^T y
# Solved efficiently using: np.linalg.lstsq(X, y, rcond=None)
#
# Why Linear Regression?
#   Sales is a continuous numerical value — regression is the correct
#   ML approach when the output is not a category but a number.
# ============================================================

print('=' * 55)
print('ML MODEL 2: LINEAR REGRESSION (numpy from scratch)')
print('=' * 55)

# --- 5C.1 Prepare regression data ---
lr_df = df[['Sales', 'Quantity', 'Discount', 'Category']].dropna().copy()
# Encode Category as integer using pd.factorize()
lr_df['cat_code'] = pd.factorize(lr_df['Category'])[0]

# Build feature matrix (3 features) and target vector
X_lr_raw = np.column_stack([
    lr_df['Quantity'].to_numpy(),
    lr_df['Discount'].to_numpy(),
    lr_df['cat_code'].to_numpy()
])
y_lr = lr_df['Sales'].to_numpy()

# Add bias column (intercept term)
X_lr = np.hstack([np.ones((X_lr_raw.shape[0], 1)), X_lr_raw])

# --- 5C.2 Manual 80/20 train/test split ---
np.random.seed(42)
idx_lr   = np.random.permutation(len(X_lr))
split_lr = int(0.8 * len(X_lr))
X_lr_train = X_lr[idx_lr[:split_lr]]
X_lr_test  = X_lr[idx_lr[split_lr:]]
y_lr_train = y_lr[idx_lr[:split_lr]]
y_lr_test  = y_lr[idx_lr[split_lr:]]
print(f"Training: {len(X_lr_train)} samples | Testing: {len(X_lr_test)} samples")

# --- 5C.3 Solve using the Normal Equation ---
# np.linalg.lstsq(A, b, rcond=None) solves the system Ax = b
# in the least-squares sense — minimises total squared error.
# Returns: (coefficients, residuals, rank, singular_values)
# Syntax: coeffs, _, _, _ = np.linalg.lstsq(X_train, y_train, rcond=None)
coeffs, _, _, _ = np.linalg.lstsq(X_lr_train, y_lr_train, rcond=None)

print("\nModel Coefficients (what the model learned):")
feature_names = ['Intercept', 'Quantity', 'Discount', 'Category Code']
for name, coef in zip(feature_names, coeffs):
    print(f"  {name:20s}: {coef:.4f}")

# --- 5C.4 Predict and evaluate ---
# np.dot(X_test, coeffs) → multiplies features by learned weights → predictions
y_lr_pred = np.dot(X_lr_test, coeffs)

# MSE: Mean Squared Error — average squared difference between predicted and actual
# Syntax: np.mean((predictions - actual) ** 2)
mse  = np.mean((y_lr_pred - y_lr_test) ** 2)
rmse = np.sqrt(mse)    # Root MSE — in same units as Sales ($)

# R² Score: how much variance the model explains
# Formula: 1 - (sum of squared residuals / total variance)
# 1.0 = perfect fit | 0.0 = no better than guessing the mean
ss_res = np.sum((y_lr_pred - y_lr_test) ** 2)
ss_tot = np.sum((y_lr_test - np.mean(y_lr_test)) ** 2)
r2     = 1 - (ss_res / ss_tot)

print(f"\nR² Score : {r2:.4f}   (model explains {r2*100:.1f}% of Sales variance)")
print(f"RMSE     : ${rmse:.2f}   (average prediction error in dollars)")

print("""
Business Interpretation:
  The model estimates expected Sales based on Quantity, Discount, and Category.
  Business uses:
  1. Forecast revenue from incoming orders before processing
  2. Set realistic sales targets per category and region
  3. Understand how discount depth affects expected revenue
""")
````

````python
# ============================================================
# STEP 5D: VISUALISE — Actual vs Predicted Sales
# Purpose: Visually assess how accurate the Linear Regression
#          predictions are against real Sales values.
# ============================================================

fig, ax = plt.subplots(figsize=(8, 6))

# Scatter plot: each dot represents one test order
# x-axis = actual Sales value, y-axis = model predicted value
# alpha=0.35 makes dots semi-transparent to reveal density
ax.scatter(y_lr_test, y_lr_pred, alpha=0.35, color='steelblue', label='Predictions')

# Perfect fit reference line (45 degrees)
# Any dot ON this line = the model predicted exactly right
mn = min(y_lr_test.min(), y_lr_pred.min())
mx = max(y_lr_test.max(), y_lr_pred.max())
ax.plot([mn, mx], [mn, mx], 'r--', lw=2, label='Perfect fit')

ax.set_title('Linear Regression: Actual vs Predicted Sales', fontsize=13, fontweight='bold')
ax.set_xlabel('Actual Sales ($)')
ax.set_ylabel('Predicted Sales ($)')
ax.legend()
plt.tight_layout()
plt.show()

print(f"Insight: R² = {r2:.4f} — the model explains {r2*100:.1f}% of the variance in Sales.")
print("  Adding features like Region or Ship Mode could improve R² further.")
````
